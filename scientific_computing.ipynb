```python
x = np.linspace(-5, 5, 100)

y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("The function y = x²")

plt.show()
```

```python
y = x**2
```

```python
y1 = x**2
y2 = 2*x + 3

plt.plot(x, y1, label="x²")
plt.plot(x, y2, label="2x + 3")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Two mathematical functions")

plt.legend()
plt.show()
```

```python
dy_dx = np.gradient(y1, x)

plt.plot(x, y1, label="y = x²")
plt.plot(x, dy_dx, label="dy/dx = 2x")

plt.xlabel("x")
plt.ylabel("value")
plt.title("Function and its derivative")

plt.legend()
plt.show()
```

```python
area = np.trapezoid(y1, x)

area
```

```python
plt.plot(x, y1)
plt.fill_between(x, y1, alpha=0.3)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Area under y = x²")

plt.show()
```

```python
x = np.linspace(0, 5, 100)

dy_dx = 2*x

y = np.cumsum(dy_dx) * (x[1] - x[0])

plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Numerical solution")

plt.show()
```

```python
x = np.linspace(0, 5, 11)

print(x)
```

```python
plt.plot(x, np.zeros_like(x), "o-")

plt.xlabel("x")
plt.title("A simple 1D mesh")

plt.yticks([])

plt.show()
```

```python
x_fine = np.linspace(0, 5, 51)

plt.plot(x_fine, np.zeros_like(x_fine), "o-")

plt.xlabel("x")
plt.title("A finer 1D mesh")

plt.yticks([])

plt.show()
```

```python
y_mesh = x_fine**2

print(y_mesh)
```

```python
plt.plot(x_fine, y_mesh, "o-")

plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x² on a mesh")

plt.show()
```

```python
plt.plot(x_fine, y_mesh, "o-")

plt.xlim(0, 1)
plt.ylim(0, 1.1)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Mesh nodes on y = x²")

plt.grid()

plt.show()
```

```python
slope = np.diff(y_mesh) / np.diff(x_fine)

print(slope[:10])
```

```python
x_mid = (x_fine[:-1] + x_fine[1:]) / 2

exact_derivative = 2 * x_mid

plt.plot(x_mid, slope, "o-", label="Numerical slope")
plt.plot(x_mid, exact_derivative, label="Exact derivative 2x")

plt.xlabel("x")
plt.ylabel("dy/dx")
plt.title("Numerical slope vs exact derivative")

plt.legend()
plt.grid()

plt.show()
```

```python
x_coarse = np.linspace(0, 5, 6)

y_coarse = x_coarse**2

plt.plot(x_coarse, y_coarse, "o-")

plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x² on a coarse mesh")

plt.grid()
plt.show()
```
