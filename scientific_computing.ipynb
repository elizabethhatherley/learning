```python
x_fine = np.linspace(0, 5, 51)

plt.plot(x_fine, np.zeros_like(x_fine), "o-")

plt.xlabel("x")
plt.title("A finer 1D mesh")

plt.yticks([])

plt.show()
```

```python
y_mesh = x_fine**2

print(y_mesh)
```

```python
plt.plot(x_fine, y_mesh, "o-")

plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x² on a mesh")

plt.show()
```

```python
plt.plot(x_fine, y_mesh, "o-")

plt.xlim(0, 1)
plt.ylim(0, 1.1)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Mesh nodes on y = x²")

plt.grid()

plt.show()
```

```python
slope = np.diff(y_mesh) / np.diff(x_fine)

print(slope[:10])
```

```python
x_mid = (x_fine[:-1] + x_fine[1:]) / 2

exact_derivative = 2 * x_mid

plt.plot(x_mid, slope, "o-", label="Numerical slope")
plt.plot(x_mid, exact_derivative, label="Exact derivative 2x")

plt.xlabel("x")
plt.ylabel("dy/dx")
plt.title("Numerical slope vs exact derivative")

plt.legend()
plt.grid()

plt.show()
```

```python
u = X**2 + Y**2

print(u)
```

```python
plt.contourf(X, Y, u)

plt.xlabel("x")
plt.ylabel("y")
plt.title("u = x² + y²")

plt.colorbar(label="u")

plt.axis("equal")
plt.show()
```

```python
print("u[0,0] =", u[0,0])
print("u[0,1] =", u[0,1])
print("u[1,0] =", u[1,0])
print("u[1,1] =", u[1,1])
```

```python
print("u values in the bottom-left element:")
print(u[0:2, 0:2])
```

```python
x_centre = 0.1
y_centre = 0.1

u_centre = x_centre**2 + y_centre**2

print("u at the centre =", u_centre)
```

```python
u_interpolated = (
    u[0, 0] +
    u[0, 1] +
    u[1, 0] +
    u[1, 1]
) / 4

print("Interpolated value at centre =", u_interpolated)
```

```python
xi = 0.5
eta = 0.5
```

```python
u_fe = (
    (1 - xi) * (1 - eta) * u[0, 0]
    + xi * (1 - eta) * u[0, 1]
    + (1 - xi) * eta * u[1, 0]
    + xi * eta * u[1, 1]
)

print("Finite-element interpolated value =", u_fe)
```

```python
xi = 0.5
eta = 0.5

w00 = (1 - xi) * (1 - eta)
w10 = xi * (1 - eta)
w01 = (1 - xi) * eta
w11 = xi * eta

print("weight at bottom-left =", w00)
print("weight at bottom-right =", w10)
print("weight at top-left =", w01)
print("weight at top-right =", w11)
```

```python
xi = 0.25
eta = 0.5
```

```python
w00 = (1 - xi) * (1 - eta)
w10 = xi * (1 - eta)
w01 = (1 - xi) * eta
w11 = xi * eta

print("bottom-left =", w00)
print("bottom-right =", w10)
print("top-left =", w01)
print("top-right =", w11)
```

```python
u_fe = (
    w00 * u[0, 0]
    + w10 * u[0, 1]
    + w01 * u[1, 0]
    + w11 * u[1, 1]
)

print("Finite-element interpolated value =", u_fe)
```

```python
x_point = 0.05
y_point = 0.1

u_exact = x_point**2 + y_point**2

print("Exact value =", u_exact)
```

```python
error = abs(u_fe - u_exact)

print("Approximation error =", error)
```

```python
x_fine = np.linspace(0, 1, 11)
y_fine = np.linspace(0, 1, 11)

X_fine, Y_fine = np.meshgrid(x_fine, y_fine)

print("Number of nodes =", X_fine.size)
```

```python
u_fine = X_fine**2 + Y_fine**2

print(u_fine[0:2, 0:2])
```

```python
xi = 0.25
eta = 0.5
```

```python
u_fe_fine = (
    (1 - xi) * (1 - eta) * u_fine[0, 0]
    + xi * (1 - eta) * u_fine[0, 1]
    + (1 - xi) * eta * u_fine[1, 0]
    + xi * eta * u_fine[1, 1]
)

print("Fine-mesh FE value =", u_fe_fine)
```

```python
x_point_fine = 0.025
y_point_fine = 0.05

u_exact_fine = x_point_fine**2 + y_point_fine**2

print("Exact value =", u_exact_fine)

error_fine = abs(u_fe_fine - u_exact_fine)

print("Fine-mesh error =", error_fine)
```
